```python

# operadores matematicos
# + - soma
# - - subtração
# / - divisão
# % - modulo
# * - multiplicação
# = - recebe
# == - igual

# para condicionar alguma coisa  precisamos :
# comando de condicao (if)
#condicao da comparacao
# resposta
#o que fazer se  nao conseguir a resposta
# if é a resposta verdadeira
# else é a resposta falsa

#exemplo qual e o mairo numero

numero1 = int(input("Informe o numero "))
numero2 = int(input("Informe um novo numero "))

#laco de condicao
# se o numero1 for maior que o numero2 faça
if numero1 > numero2:
  print("O Numero 1 é maior que o Numero 2")
elif numero1 == numero2:
  print("Os valores são iguais")

else:
  print("O Numero 2 é maior que o Numero 1")
```

```python
#Vamos Jogar par ou impar -- ROBERTA
# Pedir para o usuario 2 numeros
# somar esses 2 numeros
# descobrir se a soma é par ou impar
# (O DESAFIO ESTA AQUI) # informar se deu par ou impar (usar if)


numero1 = int(input("Informe um numero "))
numero2 = int(input("Informe um novo numero "))

soma = numero1 + numero2
print(f"A soma dos números é: {soma}")

if soma % 2 == 0:
    print("A soma é PAR")
else:
    print("A soma é ÍMPAR")
```

```python
# 1 - pedir para o usuario dois numeros
n1 = int(input('Informe o número 1: '))
n2 = int(input('Informe o número 2: '))

# 2 - somar esses dois numeros
soma = n1 + n2
print(f'a soma deu {soma}')

# 3 - descobrir se essa soma é par ou impar
# (O DESAFIO ESTA AQUI)
resultado = soma % 2

# 4 - informar se deu par ou impar (USAR IF)
if resultado == 0:
    print("deu par")
else:
    print("deu impar")
```
